# Bibliotecas

```python
!pip install -q spacy pandas
!python -m spacy download pt_core_news_sm -q

import spacy
import pandas as pd
from spacy import displacy

nlp = spacy.load("pt_core_news_sm")

print("Ambiente configurado com sucesso.")
```

```python
import nltk


from nltk.stem import RSLPStemmer

# Baixa os arquivos necessários para o stemmer em português
nltk.download("rslp")

# Frase do exercício
frase_exercicio = "O proprietário alugou a casa para a família rapidamente."

doc = nlp(frase_exercicio)
```

# Passo 1: Tokenização

```python

tokens = [token.text for token in doc]

print("PASSO 1 - TOKENIZAÇÃO")
print(tokens)
```

# Passo 2: Remoção de pontuação e stopwords

```python
tokens_filtrados = [
    token.text
    for token in doc
    if not token.is_punct and not token.is_stop
]

print("\nPASSO 2 - SEM PONTUAÇÃO E STOPWORDS")
print(tokens_filtrados)
```

# Passo 3: Stemming e Lematização

```python

stemmer = RSLPStemmer()

dados = []

for token in doc:
    if not token.is_punct and not token.is_stop:
        dados.append({
            "Palavra": token.text,
            "Stemming": stemmer.stem(token.text.lower()),
            "Lematização": token.lemma_
        })

tabela = pd.DataFrame(dados)

print("\nPASSO 3 - STEMMING E LEMATIZAÇÃO")
display(tabela)
```

# Passo 4: POS Tagging

```python
dados_pos = []

for token in doc:
    if not token.is_punct:
        dados_pos.append({
            "Palavra": token.text,
            "POS": token.pos_,
            "Tag detalhada": token.tag_
        })

tabela_pos = pd.DataFrame(dados_pos)

print("\nPASSO 4 - POS TAGGING")
display(tabela_pos)
```

# Passo 5: Sintagma Nominal e Sintagma Verbal

```python

print("\nPASSO 5 - SINTAGMAS")

print("\nSintagmas Nominais:")
for chunk in doc.noun_chunks:
    print("-", chunk.text)

print("\nSintagmas Verbais:")
for token in doc:
    if token.pos_ == "VERB":
        palavras = sorted(
            list(token.subtree),
            key=lambda x: x.i
        )

        sintagma_verbal = " ".join(
            palavra.text for palavra in palavras
        )

        print("-", sintagma_verbal)
```

# Passo 6: Similaridade entre duas palavras

```python
palavra1 = nlp("casa")
palavra2 = nlp("família")

similaridade = palavra1.similarity(palavra2)

print("\nPASSO 6 - SIMILARIDADE")
print(
    f"Similaridade entre 'casa' e 'família': "
    f"{similaridade:.3f}"
);
```
