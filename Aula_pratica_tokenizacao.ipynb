# Setup do Ambiente

## Bibliotecas


```
!pip install -q spacy nltk pandas
!python -m spacy download pt_core_news_sm -q

import nltk
import spacy
import pandas as pd

nltk.download('punkt', quiet=True)
nltk.download('punkt_tab', quiet=True)

nlp = spacy.load("pt_core_news_sm")

print("Ambiente configurado com sucesso.")
```

## Tokenzinação


```
frase_ancora = "O banco aprovou o empréstimo ontem."

# 1.1 Tentativa com método nativo do Python (.split())
tokens_split = frase_ancora.split()
print("Tokens com .split():", tokens_split)

# Observe: o ponto final "." ficou colado na palavra "ontem."!
```

```
# 1.2 Tokenização com NLTK
from nltk.tokenize import word_tokenize

tokens_nltk = word_tokenize(frase_ancora, language='portuguese')
print("Tokens com NLTK:   ", tokens_nltk)

# 1.3 Tokenização com spaCy
doc_spacy = nlp(frase_ancora)
tokens_spacy = [token.text for token in doc_spacy]
print("Tokens com spaCy:  ", tokens_spacy)
```

```
frase_desafio = "O apartamento custa R$ 450.000,00 na Av. Paulista, nº 1000! Entrar em contato via lead@email.com."

print("--- NLTK ---")
print(word_tokenize(frase_desafio, language='portuguese'))

print("\n--- spaCy ---")
doc_desafio = nlp(frase_desafio)
print([token.text for token in doc_desafio])
```
