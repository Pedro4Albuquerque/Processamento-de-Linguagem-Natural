# Bibliotecas de recursos

```python

!pip install -q spacy pandas
!python -m spacy download pt_core_news_sm -q

import spacy
import pandas as pd
from spacy import displacy

nlp = spacy.load("pt_core_news_sm")

print("Ambiente configurado com sucesso.")
```

# Atividade de Tokenização

```python
frase_exercicio = "O corretor apresentou o imóvel para o cliente ontem."

doc_exercicio = nlp(frase_exercicio)

print([token.text for token in doc_exercicio])
```


# 1. POS Tagging

```python
print("PALAVRA\t\tPOS")
print("-" * 30)

for token in doc_exercicio:
    print(f"{token.text:<15} {token.pos_}")
```

# 2. Visualização com displacy

```python
from spacy import displacy

displacy.render(
    doc_exercicio,
    style="dep",
    jupyter=True
)
```


# 3. Extração do SN e do SV

## Encontrar o sujeito

```python
sujeito = None

for token in doc_exercicio:
    if token.dep_ in ["nsubj", "nsubj:pass"]:
        sujeito = token
        break
```

## Extrair Sintagma Nominal

```python
if sujeito:
    sn_tokens = list(sujeito.subtree)
    sn = " ".join(token.text for token in sn_tokens)
else:
    sn = "Não encontrado"
```

## Encontrar o verbo principal

```python
verbo_principal = None

for token in doc_exercicio:
    if token.dep_ == "ROOT":
        verbo_principal = token
        break
```

## Extrair Sintagma Verbal

```python
if verbo_principal:
    tokens_sujeito = set(sujeito.subtree) if sujeito else set()

    sv_tokens = [
        token for token in verbo_principal.subtree
        if token not in tokens_sujeito and not token.is_punct
    ]

    sv = " ".join(token.text for token in sv_tokens)
else:
    sv = "Não encontrado"

print("Sintagma Nominal (SN):", sn)
print("Sintagma Verbal (SV):", sv)
```
