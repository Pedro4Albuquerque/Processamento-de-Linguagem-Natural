# Atividade De Tokenização

## Bibliotecas


```
!python -m spacy download pt_core_news_sm -q

import spacy
import pandas as pd

nlp = spacy.load("pt_core_news_sm")
```

## 1. Tokenização com spaCy


Frase do Exercicio

```
frase_exercicio = "Olá! Gostaria de agendar uma visita para ver o imóvel de 3 quartos amanhã às 14h?"
```

Tokenização

```
doc = nlp(frase_exercicio)

tokens_originais = [token.text for token in doc]

print("Tokens originais:")
print(tokens_originais)
```

## 2. Remoção de pontuação (use token.is_punct)

```
tokens_sem_pontuacao = [
    token.text for token in doc
    if not token.is_punct
]

print("\nTokens sem pontuação:")
print(tokens_sem_pontuacao)
```

## 3. Contagem: original vs. sem pontuação

```
print("\nQuantidade de tokens originais:", len(tokens_originais))
print("Quantidade de tokens sem pontuação:", len(tokens_sem_pontuacao))
```
